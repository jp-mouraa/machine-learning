## Discrete Spaces

```python
env_name = "FrozenLake-v1"

if env_name == "CartPole-v1":
    actions = {0: "LEFT", 1: "RIGHT"}

elif env_name == "FrozenLake-v1":
    actions = {0: "LEFT", 1: "DOWN", 2: "RIGHT", 3: "UP"}

elif env_name == "MountainCar-v0":
    actions = {0: "LEFT", 1: "RIGHT", 2: "STOP"}

# Create the environment
env = gym.make(env_name, render_mode="human") # render_mode="human" to visualize the environment

# Start a new episode/game
observation, info = env.reset(seed=42) # There is options parameter for some environments

episode_over = False
total_reward = 0

while not episode_over:
    action = env.action_space.sample() # Choose a random action
    print(f"Action taken: {actions[action]}")

    # Take the action and see what happens
    observation, reward, terminated, truncated, info = env.step(action)
    total_reward += reward
    episode_over = terminated or truncated

    # reward: +1 if the agent reaches the goal, 0 otherwise
    # terminated: True if agent fails
    # truncated: True if we hit the time limit of steps

    print(f"Observation: {observation}, "
          f"Reward: {reward}, "
          f"Terminated: {terminated}, "
          f"Truncated: {truncated}")

print(f"Episode finished! Total reward: {total_reward}")

env.close()
```

## Box Space

```python
env_name = "CarRacing-v3"

if env_name == "CarRacing-v3":
    actions = None

# Create the environment
env = gym.make(env_name, domain_randomize=True, render_mode="human")

# Start a new episode
observation, info = env.reset(seed=42)

episode_over = False
total_reward = 0

while not episode_over:
    action = env.action_space.sample() # Choose a random action
    # action = np.array([0.0, 1.0, 0.0])  # It is possible set the action as a numpy array for continuous action spaces

    # Execute the action
    observation, reward, terminated, truncated, info = env.step(action)

    total_reward += reward
    episode_over = terminated or truncated

print(f"Episode finished! Total reward: {total_reward:.2f}")

env.close()
```
