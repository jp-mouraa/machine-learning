## 2. Solve the problem using the Policy Iteration (PI) algorithm
We are now going to implement PI. This algorithm has two steps:

We are now going to implement PI. This algorithm has two steps:
- **policy evaluation:** we first evalutate a policy by computing $V^{\pi}(s) = \mathbb{E}\Big[ \sum_{t=0}^{+\infty} \gamma^t r\big(S_t,\pi(S_t)\big) \Big]$, *i.e.* we iterate using the Bellman operator until the convergence of the value of the policy $\pi$:
$$V^{\pi}(s) = r(\pi(s),a) + \gamma \sum_{s' \in \mathcal{S}} T(s,\pi(s),s') V^{\pi}(s')$$

- **policy improvement:** we then improve the policy using this value, *i.e.* $V^{\pi_n}=V^{\pi}$, to that we improve the value such as: $$V^{\pi_{n+1}}(s) = \max_{a\in \mathcal{A}} [r(s,a) + \gamma \sum_{s' \in \mathcal{S}} T(s,a,s') V^{\pi_n}(s')]$$ and we extract the corresponding greedy policy $$\pi_{n+1}(s) = argmax_{a\in \mathcal{A}} [r(s,a) + \gamma \sum_{s' \in \mathcal{S}} T(s,a,s') V^{\pi_{n+1}}(s')]$$
